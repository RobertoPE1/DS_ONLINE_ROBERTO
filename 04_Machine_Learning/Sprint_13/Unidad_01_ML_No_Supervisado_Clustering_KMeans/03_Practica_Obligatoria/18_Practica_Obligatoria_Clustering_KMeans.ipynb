### #1.2 

Aunque no lo vas a usar hasta el final de la práctica, muestra el target. Luego cargalo todo en un mismo dataframe (tendrás que añadir el target) y obtén otro dataset con todas las imagenes reordenadas aleatoriamente (emplea por ejemplo el método `sample` del dataframe o el método que tú quieras)

````python
# Mostrar el target
y = data.target
print("Target:", y)

# Crear dataframe con las imágenes aplanadas + target
df = pd.DataFrame(data.data)
df["target"] = y

# Reordenar aleatoriamente
df_shuffled = df.sample(frac=1, random_state=42).reset_index(drop=True)

df_shuffled.head()
````

### #1.3

Vamos a dividir en train y test, pero OJO RECUERDA QUE EN LOS PROBLEMAS NO SUPERVISADOS NO HAY SPLIT (porque no hay target), aquí lo hacemos para poder comparar posteriormente el clustering con la clasificación (ya que es uan práctica formativa).

Por tanto, divídelo en un conjunto de entrenamiento, un conjunto de validación y un conjunto de pruebas (80-10-10). Dado que el conjunto de datos es bastante pequeño, emplea un muestreo estratificado para asegurarse de que haya el mismo número de imágenes por persona en cada conjunto (estratificando por la columna que contenga el target)

````python
# Primero: train (80%) y temp (20%)
train_df, temp_df = train_test_split(
    df_shuffled,
    test_size=0.20,
    stratify=df_shuffled["target"],
    random_state=42
)

# Segundo: valid (10%) y test (10%) → 50/50 del 20%
valid_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["target"],
    random_state=42
)

print("Tamaños:")
print("Train:", train_df.shape)
print("Valid:", valid_df.shape)
print("Test:", test_df.shape)
````

### #1.4

Crea los pares X,y para train, validation y test.

Utiliza la siguiente función para visualizar alguna de las caras (observa que tienes que dar la X y la y, usa iloc en ambos datasets)

````python
def plot_faces(faces, labels, n_cols=5):
    faces = faces.reshape(-1, 64, 64)
    n_rows = (len(faces) - 1) // n_cols + 1
    plt.figure(figsize=(n_cols, n_rows * 1.1))
    for index, (face, label) in enumerate(zip(faces, labels)):
        plt.subplot(n_rows, n_cols, index + 1)
        plt.imshow(face, cmap="gray")
        plt.axis("off")
        plt.title(label)
    plt.show()
````

````python
# Crear X e y
X_train = train_df.drop(columns=["target"]).values.reshape(-1, 64, 64)
y_train = train_df["target"].values

X_valid = valid_df.drop(columns=["target"]).values.reshape(-1, 64, 64)
y_valid = valid_df["target"].values

X_test = test_df.drop(columns=["target"]).values.reshape(-1, 64, 64)
y_test = test_df["target"].values
````

### #1.5

Para acelerar las cosas, reduciremos la dimensionalidad de los datos utilizando PCA (técnica que veremos en el siguiente sprint). Modifica la siguiente celda de forma que las X se correspondan con las que has utilizado en el ejercicio anterior.

````python
from sklearn.decomposition import PCA

pca = PCA(n_components=50)
X_train_pca = pca.fit_transform(train_df.drop(columns=["target"]))
X_valid_pca = pca.transform(valid_df.drop(columns=["target"]))
X_test_pca = pca.transform(test_df.drop(columns=["target"]))
````

### 1.6

Aquí viene la parte del león. A continuación, agrupa las imágenes utilizando K-Means sobre el dataset de train reducido en el ejercicio anterior. Emplea el método del máximo de score de silueta para obtener el mejor k, probando con K de 5 en 5 hasta 150. ¿Cuál es el k que proporciona el mejor score de silueta? NOTA: Emplea todas las features (no hace falta seleccionar, y ya están escaladas entre 0 y 1)

````python

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

best_k = None
best_score = -1
scores = []

for k in range(5, 151, 5):
    model = KMeans(n_clusters=k, random_state=42)
    labels = model.fit_predict(X_train_pca)
    score = silhouette_score(X_train_pca, labels)
    scores.append((k, score))
    if score > best_score:
        best_score = score
        best_k = k

print(f"Mejor k según Silhouette Score: {best_k} con score {best_score:.4f}")
````

### #1.7

Repite el ejercio anterior empleando ahora el método del codo de Inercia. ¿Sale algo más concluyente o que refuerce el anterior resultado?

````python
inertias = []

for k in range(5, 151, 5):
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(X_train_pca)
    inertias.append(model.inertia_)

plt.plot(range(5, 151, 5), inertias, marker='o')
plt.xlabel("Número de clusters (k)")
plt.ylabel("Inercia")
plt.title("Método del codo")
plt.grid(True)
plt.show()
````

### #1.8

Quédate con el k obtenido con el método del score de silueta y asigna el modelo con ese k a una variable `best_model`

````python
best_model = KMeans(n_clusters=best_k, random_state=42)
best_model.fit(X_train_pca)
````

### #2.3 EXTRA VOLUNTARIO

Busca el número de clusters k que generen un algoritmo kmeans que a su vez sus distancias a los centroides sean las features de un clasificador RandomForest y que permita al clasificador obtener el mejor rendimiento: ¿Qué rendimiento puedes alcanzar? (en terminos de Accuracy)

````python
from sklearn.cluster import KMeans
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

k_values = range(5, 151, 5)
results = []

best_k_rf = None
best_acc = -1

for k in k_values:
    # Entrenar KMeans
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_train_pca)
    
    # Transformar a distancias a centroides
    X_train_km = kmeans.transform(X_train_pca)
    X_valid_km = kmeans.transform(X_valid_pca)
    
    # Entrenar RandomForest
    rf = RandomForestClassifier(n_estimators=250, random_state=42)
    rf.fit(X_train_km, y_train)
    
    # Evaluar
    y_pred = rf.predict(X_valid_km)
    acc = accuracy_score(y_valid, y_pred)
    
    results.append((k, acc))
    
    if acc > best_acc:
        best_acc = acc
        best_k_rf = k

print(f"Mejor k para RandomForest: {best_k_rf}")
print(f"Accuracy alcanzado: {best_acc:.4f}")
````
